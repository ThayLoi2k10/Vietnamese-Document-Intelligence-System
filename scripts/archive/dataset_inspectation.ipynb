```python
import os
import glob
import pandas as pd

# 1. Tự động tìm đường dẫn
def find_parquet(rel_pattern):
    for prefix in ["", "../", "../../"]:
        files = glob.glob(os.path.join(prefix, rel_pattern))
        if files:
            return sorted(files)[0]
    raise FileNotFoundError(f"Không tìm thấy: {rel_pattern}")

rel_file = find_parquet("data/raw/legal_documents/relationships/*.parquet")
meta_file = find_parquet("data/raw/legal_documents/metadata/*.parquet")

print(f"Đang đọc: {rel_file}")
rel_df = pd.read_parquet(rel_file)
meta_df = pd.read_parquet(meta_file, columns=["id", "so_ky_hieu", "title"])
meta_dict = meta_df.set_index("id")["title"].to_dict()

# 2. Thống kê tổng quan
print("=" * 80)
print("THỐNG KÊ QUAN HỆ GIỮA CÁC VĂN BẢN (RELATIONSHIPS)")
print("=" * 80)
print(f"- Tổng số cạnh quan hệ (Edges): {len(rel_df):,}")
print(f"- Số văn bản nguồn (doc_id unique): {rel_df['doc_id'].nunique():,}")
print(f"- Số văn bản đích (other_doc_id unique): {rel_df['other_doc_id'].nunique():,}")

# 3. Phân phối các loại quan hệ
print("\n--- Phân phối loại quan hệ (Top 10) ---")
type_counts = rel_df["relationship"].value_counts()
for r_type, count in type_counts.head(10).items():
    print(f"  • {r_type:<25}: {count:,} ({count/len(rel_df)*100:.2f}%)")

# 4. Kiểm tra độ phủ ID so với metadata
meta_ids = set(meta_df["id"].astype(str))
src_in_meta = rel_df["doc_id"].astype(str).isin(meta_ids).mean() * 100
tgt_in_meta = rel_df["other_doc_id"].astype(str).isin(meta_ids).mean() * 100
print(f"\n- Tỷ lệ doc_id có trong metadata:       {src_in_meta:.2f}%")
print(f"- Tỷ lệ other_doc_id có trong metadata: {tgt_in_meta:.2f}%")

# 5. Mẫu 5 mối quan hệ cụ thể kèm tiêu đề văn bản
print("\n" + "=" * 80)
print("MẪU 5 LIÊN KẾT THỰC TẾ:")
print("=" * 80)
for idx, (_, row) in enumerate(rel_df.head(5).iterrows(), start=1):
    src_id = str(row["doc_id"])
    tgt_id = str(row["other_doc_id"])
    rel = row["relationship"]
    
    src_title = meta_dict.get(src_id, "Không có tiêu đề")
    tgt_title = meta_dict.get(tgt_id, "Không có tiêu đề")
    
    print(f"\n[{idx}] Quan hệ: [{rel}]")
    print(f"  ├── Văn bản A (doc_id: {src_id}): {src_title[:80]}...")
    print(f"  └── {rel} -> Văn bản B (other_doc_id: {tgt_id}): {tgt_title[:80]}...")
```

```python
import os
import glob
import ast
import json
import re
import numpy as np
import pandas as pd
from IPython.display import display

# 1. Tự động tìm file parquet của legal_qa
def find_parquet(rel_pattern):
    for prefix in ["", "../", "../../"]:
        files = glob.glob(os.path.join(prefix, rel_pattern))
        if files:
            return sorted(files)[0]
    raise FileNotFoundError(f"Không tìm thấy: {rel_pattern}")

qa_file = find_parquet("data/raw/legal_qa/vietnamese_legal_qa.parquet")
print(f"Đang đọc file: {qa_file}")
df_qa = pd.read_parquet(qa_file)

# 2. Bộ phân giải an toàn (Resilient Parser)
def parse_qa_robust(val):
    # Case 1: Nếu vốn đã là list hoặc numpy array
    if isinstance(val, (list, np.ndarray)):
        return list(val)
    
    if not isinstance(val, str) or not val.strip():
        return []
    
    # Case 2: Thử parse JSON chuẩn
    try:
        res = json.loads(val)
        if isinstance(res, list):
            return res
    except Exception:
        pass
        
    # Case 3: Thử ast.literal_eval thông thường
    try:
        res = ast.literal_eval(val)
        if isinstance(res, list):
            return res
    except Exception:
        pass
        
    # Case 4: Xử lý dạng Numpy String (chèn dấu phẩy giữa '}\n {' hoặc '} {')
    try:
        fixed_str = re.sub(r'}\s*\n+\s*{', '}, {', val.strip())
        fixed_str = re.sub(r'}\s+{', '}, {', fixed_str)
        res = ast.literal_eval(fixed_str)
        if isinstance(res, list):
            return res
    except Exception:
        pass
        
    # Case 5: Fallback - Tìm và bóc từng object { ... } riêng lẻ
    try:
        dict_matches = re.findall(r'\{[^{}]*\}', val, re.DOTALL)
        items = []
        for d_str in dict_matches:
            try:
                items.append(ast.literal_eval(d_str))
            except Exception:
                pass
        if items:
            return items
    except Exception:
        pass

    return []

# 3. Tiến hành bóc tách & làm phẳng bảng
total_articles = len(df_qa)
flat_qa_list = []
parse_failed = 0

for idx, row in df_qa.iterrows():
    pairs = parse_qa_robust(row["generated_qa_pairs"])
    if not pairs:
        parse_failed += 1
        continue
        
    for p in pairs:
        if isinstance(p, dict):
            flat_qa_list.append({
                "doc_name": row.get("doc_name", ""),
                "doc_type_name": row.get("doc_type_name", ""),
                "context": row.get("article_content", ""),
                "question": p.get("question", ""),
                "answer": p.get("answer", ""),
                "difficulty": p.get("difficulty", "unknown"),
                "question_type": p.get("question_type", "unknown")
            })

df_flat = pd.DataFrame(flat_qa_list)

print("=" * 80)
print("TỔNG QUAN TẬP DỮ LIỆU LEGAL QA")
print("=" * 80)
print(f"- Tổng số Điều luật gốc (Records):      {total_articles:,}")
print(f"- Số bản ghi parse lỗi:                 {parse_failed:,}")
print(f"- Tổng số cặp QA bóc tách thành công:   {len(df_flat):,}")

if not df_flat.empty:
    print(f"- Trung bình QA / Điều luật:            {len(df_flat)/total_articles:.2f} câu/Điều")

    # 4. Thống kê phân phối
    print("\n" + "-" * 35 + " PHÂN PHỐI ĐỘ KHÓ (DIFFICULTY) " + "-" * 35)
    diff_stats = df_flat["difficulty"].value_counts()
    for k, v in diff_stats.items():
        print(f"  • {k:<15}: {v:,} ({v/len(df_flat)*100:.2f}%)")

    print("\n" + "-" * 35 + " PHÂN PHỐI LOẠI CÂU HỎI (QUESTION TYPE) " + "-" * 35)
    type_stats = df_flat["question_type"].value_counts()
    for k, v in type_stats.items():
        print(f"  • {k:<15}: {v:,} ({v/len(df_flat)*100:.2f}%)")

    # 5. In 3 mẫu thực tế đối chiếu theo độ khó
    print("\n" + "=" * 80)
    print("MẪU ĐỐI CHIẾU 3 CẤP ĐỘ CÂU HỎI TRONG CÙNG 1 ĐIỀU LUẬT:")
    print("=" * 80)

    sample_doc = df_flat["doc_name"].iloc[10000]
    sample_group = df_flat[df_flat["doc_name"] == sample_doc].head(3)

    print(f"📄 Văn bản gốc: {sample_doc}")
    print(f"📖 Đoạn Context (Điều luật):\n{sample_group.iloc[0]['context']}\n")

    for idx, (_, row) in enumerate(sample_group.iterrows(), start=1):
        print(f"--- [Câu {idx}] Cấp độ: [{str(row['difficulty']).upper()}] | Dạng: [{row['question_type']}] ---")
        print(f"❓ Q: {row['question']}")
        print(f"💡 A: {row['answer']}\n")
else:
    print("⚠️ Cảnh báo: Vẫn không bóc tách được dòng nào. Hãy kiểm tra giá trị của df_qa['generated_qa_pairs'].iloc[0].")
```

Legal_retrieval

```python
import os
import glob
import json
import pandas as pd

# 1. Tự động tìm đường dẫn thư mục legal_retrieval
def find_dir(rel_pattern):
    for prefix in ["", "../", "../../"]:
        p = os.path.join(prefix, rel_pattern)
        if os.path.exists(p):
            return p
    raise FileNotFoundError(f"Không tìm thấy: {rel_pattern}")

retrieval_dir = find_dir("data/raw/legal_retrieval")

# 2. Đọc corpus và đánh chỉ số toàn cục (Global Article Index: 0 -> 59,635)
with open(os.path.join(retrieval_dir, "legal_corpus.json"), "r", encoding="utf-8") as f:
    corpus = json.load(f)

global_article_map = {}
global_idx = 0

for doc in corpus:
    doc_id = doc["id"]
    law_id = doc.get("law_id", "")
    for item in doc.get("content", []):
        aid = item.get("aid")
        text = item.get("content_Article", "")
        
        # Lưu thông tin theo chỉ số tuyến tính toàn cục
        global_article_map[global_idx] = {
            "doc_id": doc_id,
            "law_id": law_id,
            "aid": aid,
            "text": text
        }
        global_idx += 1

print(f"-> Đã lập chỉ mục phẳng thành công cho toàn bộ {len(global_article_map):,} Điều luật (Index từ 0 đến {global_idx - 1:,})")

# 3. Đọc train.json và đối chiếu
with open(os.path.join(retrieval_dir, "train.json"), "r", encoding="utf-8") as f:
    train_queries = json.load(f)

print("=" * 80)
print("ĐỐI CHIẾU CHUẨN XÁC 3 CÂU HỎI VỚI ĐIỀU LUẬT CĂN CỨ (GROUND TRUTH):")
print("=" * 80)

for idx, q in enumerate(train_queries[:3], start=1):
    qid = q.get("qid")
    question = q.get("question")
    rel_laws = q.get("relevant_laws", [])
    
    print(f"\n[MẪU {idx}] QID: {qid}")
    print(f"❓ Câu hỏi: {question}")
    print(f"🎯 Số lượng Điều luật căn cứ: {len(rel_laws)}")
    
    for art_global_id in rel_laws:
        art_info = global_article_map.get(art_global_id, None)
        if art_info:
            print(f"   ├── ✅ [Global Index: {art_global_id}] | Văn bản: {art_info['law_id']} (doc_id: {art_info['doc_id']}) | Điều số (aid): {art_info['aid']}")
            print(f"   │    📖 Nội dung Điều luật: {art_info['text'][:160]}...")
        else:
            print(f"   ├── ❌ Không tìm thấy Global Index {art_global_id}")
    print("-" * 80)
```
